```python
import pandas as pd
import numpy as np

# Creating some fake ride data
data = {
    'pickup_time': ['2024-01-15 08:00', '2024-01-15 09:00', '2024-01-15 17:00'],
    'location': ['Sandton', 'Rosebank', 'Sandton'],
    'rides_count': [50, 30, 150]
}

# Converting to DataFrame
df = pd.DataFrame(data)

# Show it
print("Our first ride data:")
print(df)

# Basic analysis
print("\nTotal rides:", df['rides_count'].sum())
print("Average rides per hour:", df['rides_count'].mean())
print("Busiest location:", df.loc[df['rides_count'].idxmax(), 'location'])
```
