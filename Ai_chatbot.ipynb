```python
# Load and split the PDF
loader = PyPDFLoader("/content/the_nestle_hr_policy_pdf_2012.pdf")  # Replace with your file name
documents = loader.load()

# Split text into manageable chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
docs = text_splitter.split_documents(documents)
```

```python
# Create a vector store
# Create embeddings using the installed langchain-openai
openai_embed = OpenAIEmbeddings()
vectorstore = Chroma.from_documents(docs, openai_embed)

# Create a retrieval QA chain
qa_chain = RetrievalQA.from_chain_type(
    llm=ChatOpenAI(temperature=0),
    chain_type="stuff",
    retriever=vectorstore.as_retriever(),
)
```

```python
#Gradio Chatbot Interface
# Define chatbot interface function
def chat_with_hr_bot(query):
    result = qa_chain({"query": query})
    return result["result"]

# Create Gradio UI
interface = gr.Interface(
    fn=chat_with_hr_bot,
    inputs=gr.Textbox(label="Ask a question about Nestlé's HR Policy"),
    outputs=gr.Textbox(label="Answer from HR Assistant"),
    title="Nestlé HR Assistant Chatbot",
    description="Ask any question about Nestlé's HR policies. The assistant will respond using information extracted from the official documents."
)

interface.launch()
```
